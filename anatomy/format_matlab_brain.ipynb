```python
import numpy as np
```

```python
import sys
sys.path.append("../utils/")
from load_matlab_data import loadmat_sbx
```

```python
''' set paths '''
data_folder = f"../data/brain_surface/"
file = 'chickadee_surface_matlab.mat'
```

```python
d = loadmat_sbx(f'{data_folder}{file}')
data = d['D']
data.keys()
```

```python
# in the brain (>0.6) or hippocampus (>0.4)
# voxels: yy (ml), xx (ap), zz (dv)
brain = data['Ibrain'] # shape (n_y, n_x, n_z)
hpc = data['Ihippo'] # shape (n_y, n_x, n_z)

# convert voxels to mm
# ap is relative to lambda, ml relative to midline, dv is internally correct but 0 is arbitrary...
ap = data['xx'] # shape (n_voxels,)
ml = data['yy'] # shape (n_voxels,)
dv = data['zz'] # shape (n_voxels,)
```

```python
surface = data['brainSurface']
```

```python
# zero index the voxels
surface['faces'] = surface['faces'] - 1
surface['vertices'] = surface['vertices'] - 1
```

```python
mm = {
    'ml':ml,
    'ap':ap,
    'dv':dv
}

new_data = {
    'brain':brain,
    'hpc':hpc,
    'mm':mm,
    'surface':surface
}
```

```python
new_data['surface'].keys()
```

```python
np.save(f'{data_folder}chickadee_brain.npy', new_data)
```
