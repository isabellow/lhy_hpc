```python
import numpy as np
```

```python
'''
Take AP, ML, DV coords at a given head angle
Convert to AP, ML, DV coords at a new head angle

Here we have:
theta : original bite bar angle
u1 : AP / cos(theta)
u2 : ML
u3 : DV
B1 : [[cos(theta), 0, 0], [0, 1, 0], [sin(theta), 0, 1]]

v = B1 @ u <-- vector describing the target relative to the brain

phi : new bite bar angle
w1 : AP / cos(phi)
w2 : ML
w3 : DV
B2 : [[cos(phi), 0, 0], [0, 1, 0], [sin(phi), 0, 1]]
v = B2 @ w = B1 @ u <-- vector in terms of the new coords

w = inv(B2) @ B1 @ u <-- new coords (don't forget to compute AP from w1)
'''
```

```python
def deg_to_rad(x):
    return x * (np.pi / 180)

def rad_to_deg(x):
    return x * (180 / np.pi)
```

```python
# known coordinates
theta = deg_to_rad(-50)
ap = 0.9
ml = 0.5
dv = 6.1

# target angle
phi = deg_to_rad(-45)
```

```python
# find original v
u = np.asarray([(ap / np.cos(theta)), ml, dv])
# u = np.asarray([ap, ml, dv])
B1 = np.asarray([[np.cos(theta), 0, 0], 
                 [0, 1, 0], 
                 [np.sin(theta), 0, 1]])
v = B1 @ u
print(v)
```

```python
# compute new coords
B2 = np.asarray([[np.cos(phi), 0, 0], 
                 [0, 1, 0], 
                 [np.sin(phi), 0, 1]])
w = np.linalg.inv(B2) @ v

ap_new = w[0] * np.cos(phi)
ml_new = w[1]
dv_new = w[2]
```

```python
print(f'for beak bar at {rad_to_deg(phi)} deg.')
print(f'new AP = {ap_new:.3}')
print(f'new ML = {ml_new:.3}')
print(f'new DV = {dv_new:.3}')
```
